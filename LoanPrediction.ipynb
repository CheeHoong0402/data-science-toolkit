```python
import pandas as pd

df = pd.read_csv('/content/loan_train.csv')
display(df)
```

```python
# 2. Evaluate if the data is good or not. Does it have enough samples?

print("Rows and columns: ", df.shape)
print("Repeated loan ID: ", df['Loan_ID'].duplicated().sum())
print("Loan status(N = Unsuccessful, Y = Successful)")
print(df['Loan_Status'].value_counts())
```

```python
print("All colums: ")
print(df.columns.tolist())
print("Number of columns:", len(df.columns))
print("Possible input columns (excluding Loan_Status):", len(df.columns) - 1)
```

```python
print("Number of applicants: ", len(df))
```

```python
print(df.dtypes)
```

```python
missing = df.isnull().sum()
print("Missing values in each columns")
print(missing)
print("\nColumns with any missing values:")
print(missing[missing > 0])
print("\nNumber of columns with missing values:", (missing > 0).sum())
```

```python
target = "Loan_Status"
features = df.drop(columns = ["Loan_Status", "Loan_ID"])

print("Target: ", target)
print("Example features columns: ", features.columns.tolist())
print("Target values:", df[target].unique())
```

```python
rows_missing = df[df.isnull().any(axis=1)]
print("Number of rows with at least one missing value:", len(rows_missing))
print("First 10 rows containing missing values:")
display(rows_missing.head(10))

for column in rows_missing.columns:
    for row in rows_missing.index[rows_missing[column].isnull()]:
        print(f"Row: {row}, Column: {column}")
```
